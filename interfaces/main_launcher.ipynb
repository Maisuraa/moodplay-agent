```python
from agents.mood_agent import mood_recommendation_agent
```

```python
def run_cli():
    print("\n🎧 MoodPlay AI Agent 🎧")
    print("Talk to the agent. Type 'quit' to exit.\n")
    
    while True:
        user_input = input("You: ")
        if user_input.lower() in ["exit", "quit"]:
            print("Goodbye!")
            break

        result = mood_recommendation_agent(user_input)
        print("\nMoodPlay Agent: ")
        print(result.get("reply", "No reply generated."))
        print("\n---\n")

if __name__ == "__main__":
    run_cli()
```
