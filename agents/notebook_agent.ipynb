```python
import os
import google.generativeai as genai
```

```python
# Configure Gemini with API key
try:
    from kaggle_secrets import UserSecretsClient
    GOOGLE_API_KEY = UserSecretsClient().get_secret("GOOGLE_API_KEY")
    genai.configure(api_key=GOOGLE_API_KEY)
    print("✅ Gemini API key setup complete.")
except Exception as e:
    print(f"🔑 Authentication Error: {e}")
```

```python
# Create the model instance
notebook_agent_model = genai.GenerativeModel("models/gemini-2.5-flash")
```

```python
def infer_mood_from_text(user_input: str) -> str:
    """Infer mood from longer or natural text using Gemini."""
    prompt = f"""
You're an emotional assistant. A user just said:
"{user_input}"

From this, extract the user's mood using just one lowercase word (e.g., happy, sad, romantic, calm, energetic). 
Don't explain. Just return the single mood word.
"""
    try:
        response = notebook_agent_model.generate_content(prompt)
        return response.text.strip().lower()
    except Exception as e:
        print("Notebook Agent failed to infer mood. Error:", e)
        return "calm"  # default fallback mood
```
